# 🧠 Understanding Vector Stores and Databases

### 1. What is a Vector Store?
In traditional databases, we search for exact matches (like a username or an ID). In AI, we want to search by **meaning**. A **Vector Store** is a specialized storage system that holds data represented as numbers (vectors) so we can perform **Semantic Search**.

### 2. Core Concepts
*   **Embeddings:** The process of converting text, images, or audio into a list of numbers (a vector). These numbers represent the *meaning* of the data. Words with similar meanings will have vectors that are numerically "close" to each other.
*   **Semantic Search:** Searching based on intent and context rather than keywords. Example: Searching for "fast feline" would return results about "cheetahs" even if the word "feline" isn't in the text.
*   **Indexing:** To make searching through millions of vectors fast, the store creates an index (like a book's index). It organizes vectors in a way that the system doesn't have to check every single one to find a match.

### 3. Vector Store vs. Vector Database
While the terms are often used interchangeably, there is a subtle difference:

| Feature | Vector Store (e.g., Chroma, FAISS) | Vector Database (e.g., Pinecone, Weaviate) |
| :--- | :--- | :--- |
| **Complexity** | Usually simpler, often runs locally or in-memory. | Full-fledged system with advanced features. |
| **Scalability** | Great for small to medium projects/prototypes. | Built for massive scale and production apps. |
| **Management** | Manual handling of persistence and metadata. | Built-in data versioning, backups, and security. |

### 4. The General Workflow
1.  **Load:** Import your documents (PDFs, Text, etc.).
2.  **Split:** Break long text into smaller "chunks".
3.  **Embed:** Convert those chunks into vectors using a model (like OpenAI or HuggingFace).
4.  **Store:** Save these vectors into your **Vector Store** (like ChromaDB).
5.  **Query:** When a user asks a question, embed the question and find the most similar vectors in the store.

````python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from dotenv import load_dotenv

load_dotenv()
````

````python
#using chroma vector store
from langchain_community.vectorstores import Chroma
````

````python
from langchain_core.documents import Document


# Create LangChain documents for IPL players

doc1 = Document(
        page_content="Virat Kohli is one of the most successful and consistent batsmen in IPL history. Known for his aggressive batting style and fitness, he has led the Royal Challengers Bangalore in multiple seasons.",
        metadata={"team": "Royal Challengers Bangalore"}
    )
doc2 = Document(
        page_content="Rohit Sharma is the most successful captain in IPL history, leading Mumbai Indians to five titles. He's known for his calm demeanor and ability to play big innings under pressure.",
        metadata={"team": "Mumbai Indians"}
    )
doc3 = Document(
        page_content="MS Dhoni, famously known as Captain Cool, has led Chennai Super Kings to multiple IPL titles. His finishing skills, wicketkeeping, and leadership are legendary.",
        metadata={"team": "Chennai Super Kings"}
    )
doc4 = Document(
        page_content="Jasprit Bumrah is considered one of the best fast bowlers in T20 cricket. Playing for Mumbai Indians, he is known for his yorkers and death-over expertise.",
        metadata={"team": "Mumbai Indians"}
    )
doc5 = Document(
        page_content="Ravindra Jadeja is a dynamic all-rounder who contributes with both bat and ball. Representing Chennai Super Kings, his quick fielding and match-winning performances make him a key player.",
        metadata={"team": "Chennai Super Kings"}
    )
````

````python
embedding = GoogleGenerativeAIEmbeddings(model='gemini-embedding-001', dimensions=32)

vector_store = Chroma(
    embedding_function=embedding,
    persist_directory='my_chroma_db',
    collection_name='sample'
)
````

### Comprehensive Guide to Chroma Parameters

Here is a list of the key parameters available when initializing the `Chroma` class in LangChain:

*   **`embedding_function`**: (Required) The model used to convert text into vectors. It ensures your queries and documents are in the same 'language' for search.
*   **`persist_directory`**: (Optional) The local folder where the database is saved. If omitted, the database runs in memory and is lost when the session ends.
*   **`collection_name`**: (Optional) Defaults to 'langchain'. It allows you to partition different data sets within the same directory.
*   **`client_settings`**: (Optional) A `chromadb.config.Settings` object used for advanced configuration, such as connecting to a remote Chroma server or changing the database engine.
*   **`collection_metadata`**: (Optional) A dictionary of configuration options for the underlying Chroma collection (e.g., changing the distance metric like `{"hnsw:space": "cosine"}`).
*   **`client`**: (Optional) An existing Chroma `API` client. Use this if you have already initialized a client and want LangChain to reuse it.
*   **`relevance_score_fn`**: (Optional) A custom function to calculate similarity scores (between 0 and 1) if the default distance-to-score conversion doesn't suit your needs.

**Example with Advanced Parameters:**
```python
# Example of using collection_metadata to change search logic
vector_store = Chroma(
    embedding_function=embeddings,
    persist_directory="./chroma_db",
    collection_name="advanced_collection",
    collection_metadata={"hnsw:space": "cosine"} # Uses Cosine Similarity instead of L2 distance
)
```

````python
docs = [doc1, doc2, doc3, doc4, doc5]

# add documents
vector_store.add_documents(docs)
````

````python
# view documents
vector_store.get(include=['embeddings','documents', 'metadatas'])
````

````python
# search with similarity score
vector_store.similarity_search(
    query='Who among these are a bowler?',
    k=2 #to get top 2 result that are most close to the query
)
````

````python
# meta-data filtering (search based on meta data, we can add both query and metadata together to make search fast)
vector_store.similarity_search_with_score(
    query="Who among these are a bowler?",
    filter={"team": "Chennai Super Kings"}
)


# similarity_search_with_score, we use this to also know the score else simply we can use similarity_search
````

````python
#if we simply want data based on metadata filter we can use:
csk_players = vector_store.get(
    where={"team": "Chennai Super Kings"}
)

print("Players in Chennai Super Kings:")
for doc in csk_players['documents']:
    print(f"- {doc}")
````
