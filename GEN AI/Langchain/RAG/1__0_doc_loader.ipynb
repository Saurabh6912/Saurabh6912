```python
# Document Loader in Langchain
# This script demonstrates how to use a document loader in Langchain to load and process documents for natural language processing tasks.

# We have various of document loader available in Langchain, such as:
# - TextLoader: Loads plain text files.
# - PDFLoader: Loads PDF files. in this various of PDF loader available, such as PyPDFLoader, UnstructuredPDFLoader, etc.
    # PyPDFLoader: Uses the PyPDF2 library to load PDF files. (using when have simple text in pdf)
    # UnstructuredPDFLoader: Uses the unstructured library to load PDF files. (using when have complex layout in pdf, such as tables, images, etc.)
# - CSVLoader: Loads CSV files.
# - JSONLoader: Loads JSON files.
# - WebBaseLoader: Loads web pages.
# - DirectoryLoader: Loads all files in a directory. for example, to load all pdf from a directory using /***.pdf.
    # Example of file structure:
    # src/document_loaders/example_data/example/
    # ├── example.json
    # ├── example.jsonl
    # ├── example.txt
    # └── example.csv
```

**Here is the list of all document loader available in langchain**
- https://docs.langchain.com/oss/javascript/integrations/document_loaders#pdfs

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate

load_dotenv()

model = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite", temperature=0)

parser = StrOutputParser()
```

1. **Text Loader**

```python
from langchain_community.document_loaders import TextLoader
```

```python
loader = TextLoader('cricket.txt', encoding='utf-8')
docs = loader.load()
print(type(docs))
print(len(docs))
print(docs[0].page_content)
print(docs[0].metadata)

print(docs)

```

```python
prompt = PromptTemplate(
    template='Write a summary for the following poem - \n {poem}',
    input_variables=['poem']
)

chain = prompt | model | parser

result = chain.invoke({'poem': docs[0].page_content})
print(result)
```

3. **CSV Loader**

```python
from langchain_community.document_loaders import CSVLoader

loader = CSVLoader(file_path='Social_Network_Ads.csv')

docs = loader.load()

print(len(docs))
print(docs[0].metadata)
print(docs[1])
```

4. **WebBase Loader**

```python
from langchain_community.document_loaders import WebBaseLoader

promt = PromptTemplate(
    template = 'Answer the following question \n {question} from the following text - \n {text}',
    input_variables = ['question', 'text']
)

url = 'https://www.titan.co.in/product/titan-black-dial-analog-with-date-watch-for-men-1805qm04.html?lang=en_IN&plp=true&catID=watches-for-men'

loader = WebBaseLoader(url)
docs = loader.load()

chain = promt | model | parser

result = chain.invoke({'question': 'What is the product about? and how much valuable it is at that price (price should be mentioned as MRP)?', 'text': docs[0].page_content})
print(result)
```

5. **Directory Loader**

```python
from langchain_community.document_loaders import DirectoryLoader, PyPDFLoader

loader = DirectoryLoader(
    path='D:\\Gen AI\\PDF Test Folder',
    glob='*.pdf',
    loader_cls=PyPDFLoader
)

docs = loader.lazy_load()

doc = ''
for document in docs:
    doc += document.page_content
    print(document.metadata)
    print(document.page_content)

print(f"Total document length: {len(doc)}")
print(f"First 500 characters: {doc[:500]}")

# prompt = PromptTemplate(
#     template='What is this document about? \n {document}',
#     input_variables=['document']
# )

# chain = prompt | model | parser

# result = chain.invoke({'document': doc})
# print(result)
```
