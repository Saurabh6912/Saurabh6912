````python
# Text Splitters in LangChain
# Text splitters are used to break down large documents into smaller chunks that can be processed by language models. This is especially important when dealing with documents that exceed the token limits of the models. LangChain provides various text splitters that can be used to split text based on different criteria, such as character count, sentence boundaries, or custom delimiters.

#We have majorily 3 types of text splitters in LangChain:
# 1. Character Text Splitter: This splitter divides text based on a specified character count. It ensures that each chunk of text does not exceed the defined character limit, making it suitable for models with strict token limits.
# 2. Recursive Text Splitter: This splitter recursively splits text based on a hierarchy of delimiters. It allows for more flexible splitting by first trying to split on larger delimiters (like paragraphs) and then falling back to smaller ones (like sentences) if the chunks are still too large.
# 3. Custom Text Splitter: This splitter allows users to define their own splitting logic based on specific delimiters or patterns. It provides maximum flexibility for splitting text according to unique requirements.


# Here ww understand:
# 1. Length Based Text Splitter(Based on character count)
# 2. Structure Based splitter(based on paragraphs, sentence, characters etc.) 
    # In this we have various structure based splitters like -
    # a. Recursive Text Splitter
    # b. Spacy Text Splitter
    # c. NLTK Text Splitter
# 3. Semantic Based Splitter(based on meaning and context of the text)
    # In this we have various semantic based splitters like -
    # a. Cohere Text Splitter
    # b. OpenAI Text Splitter
    # These splitters use embeddings to determine the semantic similarity between chunks of text and split them accordingly.
    # uses cosine similarity to determine the semantic similarity between chunks of text and split them accordingly. This allows for more meaningful splits that preserve the context and meaning of the original text.

# Let's see how to use these text splitters in LangChain:
````

1. **Length Based Splitter**

````python
from langchain_text_splitters import CharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader('dl-curriculum.pdf')

docs = loader.load()

splitter = CharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=0,
    separator=''
)

result = splitter.split_documents(docs)

print(result[0].page_content)
````

2. **Structure Based Splitter**

- Text Structure Based

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text = '''
Space exploration has led to incredible scientific discoveries. From landing on the Moon to exploring Mars, humanity continues to push the boundaries of what’s possible beyond our planet.

These missions have not only expanded our knowledge of the universe but have also contributed to advancements in technology here on Earth. Satellite communications, GPS, and even certain medical imaging techniques trace their roots back to innovations driven by space programs.
'''

splitter = RecursiveCharacterTextSplitter(
    chunk_size = 300,
    chunk_overlap = 0
)

#perform the split
chunks = splitter.split_text(text)

print(len(chunks))
for c in chunks:
    print(c)
````

- Code Structure Based - like python, java, etc

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter, Language

code_text = '''
class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade  # Grade is a float (like 8.5 or 9.2)

    def get_details(self):
        return self.name"

    def is_passing(self):
        return self.grade >= 6.0


# Example usage
student1 = Student("Aarav", 20, 8.2)
print(student1.get_details())

if student1.is_passing():
    print("The student is passing.")
else:
    print("The student is not passing.")
'''

# Initialize splitter

splitter = RecursiveCharacterTextSplitter.from_language(
    language = Language.PYTHON,
    chunk_size = 300,
    chunk_overlap = 0
)

chunks = splitter.split_text(code_text)

print(len(chunks))

for c in chunks:
    print(c,'\n\n')
````

3. **Semantic Based Splitter**

````python
from langchain_experimental.text_splitter import SemanticChunker
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from dotenv import load_dotenv

load_dotenv()

text_splitter = SemanticChunker(
    GoogleGenerativeAIEmbeddings(model='gemini-embedding-001'),
    breakpoint_threshold_type="standard_deviation",
    breakpoint_threshold_amount=3
)

sample = """
Farmers were working hard in the fields, preparing the soil and planting seeds for the next season. The sun was bright, and the air smelled of earth and fresh grass. The Indian Premier League (IPL) is the biggest cricket league in the world. People all over the world watch the matches and cheer for their favourite teams.


Terrorism is a big danger to peace and safety. It causes harm to people and creates fear in cities and villages. When such attacks happen, they leave behind pain and sadness. To fight terrorism, we need strong laws, alert security forces, and support from people who care about peace and safety.
"""
docs = text_splitter.create_documents([sample])
print(len(docs))
print(docs)

#SemanticChunker Parameters are 
````

**How it Works (The 3-Step Process)Break into Sentences:** 
- The chunker first takes your large document and splits it into individual sentences.
- Calculate Meaning (Embeddings): It converts each sentence into a vector (a list of numbers representing its "meaning") using your provided embedding model.
- Detect Topic Shifts: It compares the meaning of adjacent sentences. If the "semantic distance" (the difference in meaning) between two consecutive sentences is too large, it assumes the topic has changed and creates a split there.

**How Parameters Decide the Split**
> The parameters act like "sensitivity dials" for how strict you want to be about topic changes.
- breakpoint_threshold_type: This sets the math method used to find shifts.
    - Percentile (Default): It looks at all shifts in your whole document and splits at the most dramatic ones (e.g., the top 5% of changes).
    - Standard Deviation: It splits wherever a shift is much larger than the "average" shift in that document.
- breakpoint_threshold_amount: This is the actual limit.
    - Higher values (e.g., 95 percentile) mean "only split if the topic is very different." This results in fewer, larger chunks.
    - Lower values (e.g., 60 percentile) mean "split even if the topic changes just a little." This results in more, smaller chunks.
    
- min_chunk_size: This is a safety floor. Even if the math says "split here," it won't split if the resulting chunk would be smaller than this number of characters
