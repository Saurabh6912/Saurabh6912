# Understanding Tools in LangChain

This document provides comprehensive notes on the concept of **Tools in LangChain**, their integration with LLMs, and how to build them. This is a fundamental part of the agentic workflow.

## 1. What is a Tool?

*   **Core Concept:** Large Language Models (LLMs) excel at reasoning and language generation but cannot inherently perform external actions (e.g., fetching live data, executing code, interacting with APIs).
*   **Definition:** A Tool is essentially a **Python function** specifically packaged so that an LLM can understand its purpose, call it, and interact with it effectively.
*   **Analogy:** If an LLM is considered the "brain" of an AI system, then Tools serve as its "hands and legs," enabling it to interact with the external world and take concrete actions.
*   **The Agentic Loop:** An AI Agent is an LLM-powered system designed to autonomously *think*, *decide*, and *take action* by leveraging these Tools.

## 2. Types of Tools

LangChain offers different categories of tools:

*   **Built-in Tools:** These are pre-built, production-ready tools provided by LangChain for common and widely used tasks.
    *   **Examples:**
        *   **DuckDuckGo Search:** For fetching current information from the web.
        *   **Wikipedia Query:** For retrieving information from Wikipedia.
        *   **Python REPL:** To execute Python code.
        *   **Shell Tool:** To execute shell commands.
        *   **Integrators:** Such as Gmail or Slack integrators for email and communication tasks.

*   **Custom Tools:** These are developed when there isn't an existing built-in solution that fits your specific business logic, proprietary APIs, or unique requirements.

## 3. Creating Custom Tools

There are three primary methods to implement custom tools in LangChain, offering varying levels of control and complexity:

### A. Using the `@tool` Decorator (Easiest)

This is the simplest and most recommended method for most use cases.

*   **Requirements:** A standard Python function, proper type hinting for its arguments and return value, and the `@tool` decorator from `langchain_core.tools`.
*   **Example:**
    ```python
    from langchain_core.tools import tool

    @tool
    def multiply(a: int, b: int) -> int:
        """Multiplies two numbers."""
        return a * b
    ```

### B. Using `StructuredTool` and Pydantic

This method is suitable when you need more rigid schema enforcement for complex input parameters.

*   **Requirements:** A Python function and a Pydantic `BaseModel` to define the input schema.
*   **Process:** Use `StructuredTool.from_function` to create the tool, passing the function, name, description, and the Pydantic `args_schema`.

### C. Inheriting from `BaseTool` (Most Advanced)

This method provides the highest level of customization, allowing for advanced features like asynchronous execution.

*   **Requirements:** Define a class that inherits from `BaseTool`, implementing the `_run` method (for synchronous execution) or `_arun` (for asynchronous execution).
*   **Input Schema:** Define an `args_schema` using a Pydantic `BaseModel` within the class.

## 4. Key Attributes of a Tool

Every tool in LangChain exposes specific metadata that is crucial for an LLM to understand how and when to use it:

*   **`name`:** A unique identifier for the tool. This is how the LLM refers to the tool.
*   **`description`:** A clear, concise documentation (often from the function's docstring) that helps the LLM decide if the tool is relevant for a given task and how to use it.
*   **`args` (or `args_schema`):** Defines the expected input parameters for the tool's function. This is typically a JSON schema that the LLM uses to correctly format its calls to the tool.
    *   **Important:** When an LLM calls a tool, it interacts with the **Tool Schema (JSON schema)**, not directly with the raw Python function code.

## 5. Toolkits

*   **Definition:** A `Toolkit` is a logical collection or grouping of related tools that are packaged together.
*   **Purpose:** It promotes reusability and simplifies the management of tools that serve a common purpose.
*   **Example:** A "Google Drive Toolkit" might contain tools for various operations like reading files, writing files, listing directories, etc., all related to Google Drive functionality.
*   **Benefit:** Toolkits streamline development by providing an organized way to deploy sets of tools across different applications.

**Custom Tools**

**1. Using @Tool Decorator**

````python
from langchain_core.tools import tool
````

````python
# Step 1 - create a function

def multiply(a, b):
    """Multiply two numbers"""
    return a*b
````

````python
# Step 2 - add type hints

def multiply(a: int, b:int) -> int:
    """Multiply two numbers"""
    return a*b
````

````python
# Step 3 - add tool decorator

@tool
def multiply(a: int, b:int) -> int:
    """Multiply two numbers"""
    return a*b
````

````python
result = multiply.invoke({"a":3, "b":5})
print(result)
````

````python
print(multiply.name)
print(multiply.description)
print(multiply.args)
````

````python
print(multiply.args_schema.model_json_schema())
````

**2. Using Structure Tool** - For Rigid Validation against input/output

````python
# Change this:
# from langchain.tools import StructuredTool

# To this:
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
````

````python
#step 1 - Define the base class
class MultiplyInput(BaseModel):
    a: int = Field(description="The first number to add")
    b: int = Field(description="The second number to add")
````

````python
#step 2 - Define the function
def multiply_func(a: int, b: int) -> int:
    return a * b
````

````python
#step 3- Rap the function with structured base class schema to form StructuredTool
multiply_tool = StructuredTool.from_function(
    func=multiply_func,
    name="multiply",
    description="Multiply two numbers",
    args_schema=MultiplyInput
)
````

````python
#step 4 - Tool is ready, lets invoke
result = multiply_tool.invoke({'a':3, 'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args)
````

**3. Using BaseTool class**  Inherently all the tool are the coming from BaseTool only

````python
from langchain.tools import BaseTool
from typing import Type
````

````python
#step 1 - Define the base class
# arg schema using pydantic

class MultiplyInput(BaseModel):
    a: int = Field(description="The first number to add")
    b: int = Field(description="The second number to add")
````

````python
#step 2 -Define the tool by inheriting BaseTool and implementing the _run(synchronous) or _arun(asynchronous) method

class MultiplyTool(BaseTool):
    name: str = "multiply"
    description: str = "Multiply two numbers"

    args_schema: Type[BaseModel] = MultiplyInput

    def _run(self, a: int, b: int) -> int:
        return a * b
````

````python
#step 3 - Make the object of the tool
multiply_tool = MultiplyTool()
````

````python
#step 4 - Tool is ready, lets invoke
result = multiply_tool.invoke({'a':3, 'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)

print(multiply_tool.args)


#in real all the tool is inherited from the BaseTool class only, whether  its @tool decorator or structured tool or any other built-in/custom tool
````

### **Toolkit**

````python
from langchain_core.tools import tool

#step 1 - create a tools
# Custom tools
@tool
def add(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
````

````python
#step 2 - rap the similar tools in a class to make toolkit
class MathToolkit:
    def get_tools(self):
        return [add, multiply]
````

````python
#make a toolkit object and get the tools from it
toolkit = MathToolkit()
tools = toolkit.get_tools()

for tool in tools:
    print(tool.name, "=>", tool.description)
````
