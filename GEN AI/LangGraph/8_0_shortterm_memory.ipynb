### **ShortTerm Memory Implementation using InMemorySaver - Local RAM**

```python
from langgraph.graph import StateGraph, START, MessagesState
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

```python
# Define a function to call the model with the current state
def call_model(state: MessagesState):
    response = model.invoke(state["messages"])
    return {"messages": [response]}
```

```python
#build graph with node and edge
builder = StateGraph(MessagesState)
builder.add_node("call_model", call_model)
builder.add_edge(START, "call_model")
```

```python
graph = builder.compile()
graph
```

```python
graph.invoke({"messages":[{"role": "user", "content": "Hi, I am Saurabh!"}]})
```

```python
graph.invoke({"messages":[{"role": "user", "content": "What is my name?"}]})
```

*Model not have memory to have past conversation, see saying I don't have access*

**Now Let's Implement InMemory Saver for Memory**

```python
from langgraph.checkpoint.memory import InMemorySaver

checkpointer = InMemorySaver()
graph = builder.compile(checkpointer=checkpointer)
```

```python
config = {"configurable": {"thread_id": "thread-1"}}
config2 = {"configurable": {"thread_id": "thread-2"}}
```

```python
graph.invoke({"messages":[{"role": "user", "content": "Hi, I am Saurabh!"}]},config=config)
```

```python
graph.invoke({"messages": [{"role": "user", "content": "What is my name?"}]}, config)
```

```python
graph.invoke({"messages": [{"role": "user", "content": "What is my name?"}]}, config2)
```

```python
snap = graph.get_state(config)
vals = snap.values
for m in vals.get("messages", []):
        print("-", type(m).__name__, ":", m.content)
```
